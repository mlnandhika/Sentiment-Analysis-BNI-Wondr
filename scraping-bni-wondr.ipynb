```python
# Install library
!pip install google-play-scraper

import pandas as pd
from google_play_scraper import reviews, Sort
```

```python
TARGET_REVIEWS = 17000    # jumlah review yang ingin diambil

all_reviews = []
continuation_token = None

print(f"Mulai mengambil {TARGET_REVIEWS} review...")

while len(all_reviews) < TARGET_REVIEWS:
    try:
        result, continuation_token = reviews(
            "id.bni.wondr",
            lang='id',          # bahasa Indonesia
            country='id',       # negara Indonesia
            sort=Sort.NEWEST,   # review terbaru
            count=1000,          # maksimal per request
            continuation_token=continuation_token
        )

        if not result:
            print("Tidak ada review tambahan.")
            break

        all_reviews.extend(result)

        print(f"Total review terkumpul: {len(all_reviews)}")

        if continuation_token is None:
            print("Semua review sudah diambil.")
            break

    except Exception as e:
        print("Error:", e)
        break
```

```python
# Konversi ke dataframe
df = pd.DataFrame(all_reviews)
df.head()
```

```python
df.info()
```

```python
# Simpan ke Csv
filename = "wondr_bni_reviews.csv"

df.to_csv(filename, index=False, encoding='utf-8-sig')

print("Total data:", len(df))
print("File CSV :", filename)
```
